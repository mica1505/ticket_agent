```
CREATE OR REPLACE API INTEGRATION github_integration
  API_PROVIDER = git_https_api
  API_ALLOWED_PREFIXES = ('https://github.com/mica1505')
  ENABLED = TRUE;
```

```
CREATE OR REPLACE SECRET github_token
  TYPE = PASSWORD
  USERNAME = 'mica1505'
  PASSWORD = '';
```

```
ALTER API INTEGRATION github_integration 
SET ALLOWED_AUTHENTICATION_SECRETS = ('github_token');
```

```
CREATE OR REPLACE GIT REPOSITORY ticket_agent_repo
  API_INTEGRATION = github_integration
  ORIGIN = 'https://github.com/mica1505/ticket_agent.git'
  GIT_CREDENTIALS = github_token; 
```

```
-- Récupère la dernière version du code depuis GitHub
ALTER GIT REPOSITORY ticket_agent_repo FETCH;
```
